### MNIST 손글씨 숫자 데이터셋 이해하기

<center> <img src="https://www.kaggle.com/competitions/3004/images/header" width=400></center>



- MNIST 데이터셋 : 미국 국립표준기술원(NIST)이 고등학생과 인구조사국 직원 등이 쓴 손글씨를 이용해 만든 데이터
- 70,000개의 숫
자 이미지에 각각 0~9까지 레이블를 붙인 데이터셋
- 학습 데이터 60,000개, 테스트 데이터 10,000개로 구성
- 28 * 28 크기 이미지

<center> <img src="https://learnmachinelearning.files.wordpress.com/2017/08/mnist.png?w=638" height="150px"> </center>

데이터 세트 불러오기
- load_data() : X_train, y_train, X_test, y_test 네 개의 넘파이 배열 반환
- X_train, X_test는 28×28 픽셀의 각 손글씨 이미지 데이터
- y_train, y_test는 0~9 사이의 레이블

```python
import tensorflow as tf
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()
```

데이터 타입 확인

```python
print(type(X_train))
print(type(X_test))
print(type(y_train))
print(type(y_test))
```

데이터 모양 확인

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

- 각 이미지는 28x28 픽셀 크기의 흑백(그레이스케일) 이미지이며,
- 각 픽셀은 0(검은색)부터 255(흰색)까지의 밝기 값을 가진다.
- 즉, 이미지 한 장은 784개의 숫자(픽셀 값)로 이루어진 2차원 배열이다.

## (실습) Fashion MNIST 데이터셋 이해하기

<center><img src="https://drive.google.com/uc?id=1AOEqdRP2unXX6Ab8lx3LdRqdsu1Yz4sT" width=50%></center>

```python
from keras.datasets import fashion_mnist
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()
```

훈련데이터와 테스트데이터 모양 확인

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

인덱스 1번 이미지 확인

```python
for i in X_train[1] :
  for j in i :
    print('%3d' %j, end=' ')
  print()
```

인덱스 1번 이미지의 레이블 확인

```python
print(y_train[1])
```

```python
X_train[1]
```

```python
#클래스 이름
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']
```

인덱스 1번 이미지의 클래스 확인

```python
print(class_names[y_train[1]])
```

인덱스 100번 이미지 확인

```python
X_train[100]
```

인덱스 100번 이미지의 레이블 확인

```python
print(y_train[100])
```

인덱스 100번 이미지의 클래스 확인

```python
print(class_names[y_train[100]])
```

- 클래스별 이미지가 각각 몇 개 있는지 확인해보자.

```python
for i in range(10) :
  print(class_names[i], sum(y_train == i))
```
