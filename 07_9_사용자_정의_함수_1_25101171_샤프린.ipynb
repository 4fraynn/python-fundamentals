함수 : 재사용 가능한 프로그램의 조각,
특정 블록의 명령어들을 묶어 이름을 붙이고 그 이름을 프로그램 어디에서든 사용함으로써
그 블록에 포함된 명령어들을 몇 번이고 다시 실행

- sum(), len(), max(), min(), count(), index() 등의 내장 함수를 사용하지 않고,
- 이와 동일한 기능을 수행하는 함수를 직접 설계하고 구현하는 과정을 통해
- 내장 함수의 원리 이해 및 사용자 정의 함수 설계/작성 연습

# 사용자 정의 함수


```python
#a, b 두 수의 합계를 구하는 함수
def hap(n, m ) :
  return n + m
a, b = map(int,input().split())
print(hap(a, b))
```

```python
#1부터 n까지 합계를 구하는 함수
def hap(n) :
  total = 0
  for i in range(1, n + 1):
    total += i
  return total

print(hap(int(input())))
```

```python
#a, b 두 수 사이의 정수를 리스트로 만드는 함수
def f(a, b):
  res = []
  for i in range(a, b + 1):
    res.append(i)
  return res
a, b = map(int, input().split())
c = f(a, b)
print(c)
```

```python
#a, b 두 수 사이의 홀수를 리스트로 만드는 함수
def f_odd(a, b):
  res = []
  for i in range(a, b + 1):
    if i % 2 != 0:
      res.append(i)
  return res
a, b = map(int, input().split())
c = f_odd(a, b)
print(c)
```

```python
#문자열 길이 구하는 함수
def length(s):
  cnt = 0
  for _ in s:
    cnt += 1
  return cnt
a = input('문자열 입력 : ')
res = length(a)
print('문자열 길이 : ', res)
```

```python
#리스트의 길이 구하는 함수
def length(l):
  cnt = 0
  for _ in l:
    cnt += 1
  return cnt
a = input('리스트 항목 입력 : ').split()
res = length(a)
print('리스트 길이 : ', res)
```

```python
#리스트 a에서 특정 항목 n의 갯수 구하는 함수
def cnt(n, a):
  cnt = 0
  for i in a:
    if i == n:
      cnt += 1
  return cnt
a = [9, 7, 10, 8, 5, 7, 7, 9, 3, 5]
n = int(input())
res = cnt(n, a)
print(res)
```

```python
#리스트의 특정 항목의 위치 구하는 함수, 해당 항목이 2번 이상 등장하는 경우 처음 등장하는 위치
def idx(n, a):
  i = 0
  for x in a:
    if x == n:
      return i
    i += 1

a = [9, 7, 10, 8, 5, 7, 7, 9, 3, 5]
n = int(input())
res = idx(n, a)
print(res)
```

```python
#리스트를 거꾸로 출력하는 함수
def rev(a):
  for x in a[::-1]:
    print(x)
a = input().split()
rev(a)
```
